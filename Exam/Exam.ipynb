### 3.1 Tích hợp dữ liệu thành một dataset

### 3.2 Kiểm tra và xử lý dữ liệu thiếu trong tập dữ liệu (nếu có)
---
Thực hiện kiểm tra xem trong Data thu thập có thuộc tính nào không có giá trị (missing) hay không? Nếu có phải xử lý những dòng này để đảm bảo tập dữ liệu đạt chất lượng.

```python
#Đọc file dữ liệu tổng hợp để chuẩn bị
Data_2024 = pd.read_csv('All_Data_2024.csv')
Data_2024.info()
```

```python
#Quan sát dữ liệu:
Data_2024.head()
```

```python
#Kiểm tra số lượng bản ghi null trong Dữ liệu:
#---------CODE HERE---------------------#
missing_values = Data_2024.isnull().sum()
print(missing_values)
#---------END CODE HERE-----------------#
```

```python
# Liệt kê các bản ghi chứa dữ liệu Missing:
#---------CODE HERE---------------------#
nan_df = Data_2024[Data_2024.isnull().any(axis=1)]

#---------END CODE HERE-----------------#
display(nan_df)
print('Tổng số dòng không có dữ liệu:', nan_df.shape[0])
```

```python
#--->XỬ LÝ CÁC DÒNG DỮ LIỆU THIẾU:
#Xóa tất cả các dòng dữ liệu rỗng
#---------CODE HERE---------------------#
Data_2024 = Data_2024.dropna()

#---------END CODE HERE-----------------#


#Reset lại chỉ số cột
Data_2024.reset_index(drop=True,inplace=True)

#Thông tin dữ liệu sau khi xóa các dòng rỗng:
Data_2024.info()
```

### 3.3) Kiểm tra và xử lý dữ liệu trùng lặp (nếu có)
----
Thực hiện việc Kiểm tra các bản ghi trùng lặp trong file dữ liệu

```python
#Thống kê số bản ghi trùng lặp:
Data_2024.duplicated().sum()
```

```python
#Hiển thị danh sách các bản ghi trùng nhau:
#---------CODE HERE---------------------#
duplicate_rows = Data_2024[Data_2024.duplicated()]
display(duplicate_rows)
#---------END CODE HERE-----------------#
```

Dữ liệu có nhiều bản ghi bị trùng lặp; có 2 loại trùng lặp:
1. Các dòng tiêu đề của bảng dữ liệu bị lặp lại: **Order ID -- Product -- Quantity Ordered -- Price Each -- Order Date -- Purchase Address**
2. Một số đơn hàng trong quá trình tổng hợp dữ liệu bị lặp lại

 **CÂU HỎI 1: PHƯƠNG ÁN XỬ LÝ DỮ LIỆU TRÙNG LẶP?**

trả lời ở đây : 

Loại 1. Dòng tiêu đề bị lặp lại : Do đọc 12 file tháng, mỗi file có header riêng → khi ghép lại, các dòng header bị cộng dồn vào dataset, với loại này ta chỉ cẩn xóa toàn bộ các dòng mà Order ID == "Order ID"

Loại 2. Các đơn hàng bị lặp lại : Trùng lặp bản ghi giao dịch là các dòng dữ liệu giống nhau hoàn toàn → cần loại bỏ bằng phương pháp drop_duplicates() 

```python
#1. Với các dòng tiêu đề lặp lại trong dữ liệu 
#Liệt kê các dòng tiêu đề trong Data:
header = Data_2024[Data_2024['Order ID']=='Order ID']
display(header)
print('Số dòng tiêu đề lặp lại:', header.shape[0])
```

```python
#xử lý dữ liệu trùng lặp các dòng tiêu đề lặp lại trong Data
#---------CODE HERE---------------------#
Data_2024 = Data_2024[Data_2024['Order ID'] != 'Order ID']
#---------END CODE HERE-----------------#
#Thông tin dữ liệu sau khi xử lý dòng tiêu đề lặp lại:
Data_2024.info()
```

```python
#2. Xử lý trùng lặp dữ liệu Với các đơn hàng bị lặp lại
#---------CODE HERE---------------------#
Data_2024 = Data_2024.drop_duplicates()
#---------END CODE HERE-----------------#
#Reset lại index của DataFrame:
Data_2024.reset_index(drop=True,inplace=True)

#Thông tin dữ liệu sau khi xử lý đơn hàng trùng lặp:
Data_2024.info()
```

```python
#Thống kê lại dữ liệu trùng lặp sau xử lý:
Data_2024.duplicated().sum()
```

### 3.4) Chuyển đổi kiểu dữ liệu của một số cột cho phù hợp:
---
* Dữ liệu của 2 cột 'Quantity Ordered', Price Each đang ở dạng object --> Chuyển sang kiểu dữ liệu số
* Dữ liệu của cột Order Date đang ở dạng object --> Chuyển sang dạng date time

```python
#Chuyển đổi kiểu dữ liệu sang dạng số:
Data_2024['Quantity Ordered'] = pd.to_numeric(Data_2024['Quantity Ordered'])
Data_2024['Price Each'] = pd.to_numeric(Data_2024['Price Each'])

#Chuyển đổi kiểu dữ liệu sang dạng time:
Data_2024['Order Date'] = pd.to_datetime(Data_2024['Order Date'])

Data_2024.info()
```

```python
#Sắp xếp lại dữ liệu:
Data_2024.sort_values(by='Order Date', inplace=True)
#Reset lại chỉ sổ:
Data_2024.reset_index(drop=True, inplace=True)
#Hiển thị dữ liệu:
Data_2024
```

```python
#Quan sát đặc trưng thống kê của trường Số lượng - Giá
Data_2024.describe()
```

### 3.5) Thêm các cột dữ liệu mới vào bảng để phục vụ phân tích
---
Trong nhiều trường hợp, cần bổ sung thêm các cột dữ liệu mới được tổng hợp từ các cột dữ liệu đã có để phục vụ các mục đích thống kê

#### 1. Cột Month, Hour: Month và Hour được tách ra từ cột Order Date lấy tháng, giờ mua hàng

```python
Data_2024.head()
```

```python
#Tách lấy dữ liệu Tháng,Giờ trong Order Date
#---------CODE HERE---------------------#

Data_2024['Month'] = Data_2024['Order Date'].dt.month
Data_2024['Hour'] = Data_2024['Order Date'].dt.hour
#---------END CODE HERE-----------------#

Data_2024.head()
```

```python
Data_2024.tail()
```

#### 2. cột Sales: Được tính bằng số lượng sản phẩm (Quantity Ordered) * Giá bán mỗi sản phẩm (Price Each)

```python
#Tính giá trị cột Sales
#---------CODE HERE---------------------#

Data_2024['Sales'] = Data_2024['Quantity Ordered'] * Data_2024['Price Each']
#---------END CODE HERE-----------------#
Data_2024.head()
```

#### 3.Cột City: Tách tên thành phố trong địa chỉ mua hàng vào một cột mới

```python
#Viết hàm tách lấy tên thành phố và mã bang từ địa chỉ của khách hàng:
def get_store(address):
    #---------CODE HERE---------------------#
    parts = address.split(',')         
    
    city = parts[1].strip()             
    state = parts[2].strip().split(' ')[0]  
    
    store = f"{city} ({state})"         
    #---------END CODE HERE-----------------#
    return store

#Thực hiện bổ sung thêm một thuộc tính City:
Data_2024['City'] = Data_2024['Purchase Address'].apply(lambda x: get_store(x))
Data_2024.head()
```

### Câu hỏi 4: Khách hàng thường mua các sản phầm nào cùng với nhau?
---

```python

#Lọc danh sách lấy các hóa đơn mua nhiều sản phẩm:
df = Data_2024[Data_2024['Order ID'].duplicated(keep=False)]
#Sắp xếp lại dữ liệu:
df.sort_values(by=['Order ID','Price Each'], ascending=[True, False], inplace=True)
df
```

```python
#Ghép các sản phẩm lại với nhau theo từng mã Order ID, để xem sản phẩm nào hay mua cùng với nhau
df['Grouped'] = df.groupby('Order ID')['Product'].transform(lambda x: ','.join(x))

#Chỉ hiển thị 3 cột cần thiết
df[['Order ID','Product','Grouped']]
```

```python
#Lấy cột Order ID, Grouped, xóa các hóa đơn lặp lại
df2 = df[['Order ID', 'Grouped']].drop_duplicates()
df2
```

```python
#Đếm các đơn hàng có Grouped giống nhau --> Xem các mặt hàng nào thường được mua cùng nhau:
from itertools import combinations
from collections import Counter

count = Counter()

for row in df2['Grouped']:
    row_list = row.split(',')
    count.update(Counter(combinations(row_list, 2)))

#Hiển thị 15 đơn hàng mua sản phẩm cùng nhau xuất hiện nhiều nhất trong tập dữ liệu:
for key,value in count.most_common(15):
    print(key, value)
```
