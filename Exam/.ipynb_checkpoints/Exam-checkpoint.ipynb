### Câu hỏi 4: Khách hàng thường mua các sản phầm nào cùng với nhau?
---

```python
#Lọc danh sách lấy các hóa đơn mua nhiều sản phẩm:
df = Data_2024[Data_2024['Order ID'].duplicated(keep=False)]
#Sắp xếp lại dữ liệu:
df.sort_values(by=['Order ID','Price Each'], ascending=[True, False], inplace=True)
df
```

```python
#Ghép các sản phẩm lại với nhau theo từng mã Order ID, để xem sản phẩm nào hay mua cùng với nhau
df['Grouped'] = df.groupby('Order ID')['Product'].transform(lambda x: ','.join(x))

#Chỉ hiển thị 3 cột cần thiết
df[['Order ID','Product','Grouped']]
```

```python
#Lấy cột Order ID, Grouped, xóa các hóa đơn lặp lại
df2 = df[['Order ID', 'Grouped']].drop_duplicates()
df2
```

```python
#Đếm các đơn hàng có Grouped giống nhau --> Xem các mặt hàng nào thường được mua cùng nhau:
from itertools import combinations
from collections import Counter

count = Counter()

for row in df2['Grouped']:
    row_list = row.split(',')
    count.update(Counter(combinations(row_list, 2)))

#Hiển thị 15 đơn hàng mua sản phẩm cùng nhau xuất hiện nhiều nhất trong tập dữ liệu:
for key,value in count.most_common(15):
    print(key, value)
```
