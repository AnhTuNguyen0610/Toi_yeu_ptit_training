### **Yêu cầu 1.3:**
A) Xóa tất cả các dòng dữ liệu có số lượng người vote (vote_count) dưới 100

B) Chuyển đổi kiểu dữ iệu cột release_date về kiểu datetimes, sắp xếp dữ liệu theo thứ tự giảm dần của thời gian (bộ film mới ra mắt ở trên cùng của dữ liệu)

```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}

df_movies = df_movies[df_movies['vote_count'] >= 100]
df_movies['release_date'] = pd.to_datetime(df_movies['release_date'], errors='coerce')
df_movies.sort_values(by='release_date', ascending=False, inplace=True)

print(df_movies.info())
```

### **Yêu cầu 1.4:**

A) Hiển thị danh sách 10 bộ film có số lượng người đánh giá cao nhất (vote_cout).

B) Hiển thị danh sách 10 bộ film có điểm đánh giá trung bình cao nhất (vote_average)

```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}


top_10_vote_count = df_movies.sort_values(by='vote_count', ascending=False).head(10)
print("10 bộ phim có số lượng người đánh giá cao nhất (vote_count):")
print(top_10_vote_count[['original_title', 'vote_count']])

top_10_vote_average = df_movies.sort_values(by='vote_average', ascending=False).head(10)
print("\n10 bộ phim có điểm đánh giá trung bình cao nhất (vote_average):")
print(top_10_vote_average[['original_title', 'vote_average']])


```

### **Yêu cầu 1.5:**
A) Nhóm dữ liệu theo thuộc tính adult, đếm số lượng bộ phim và tổng số lượt vote:


```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}

grouped = df_movies.groupby('adult').agg(
    total_movies=('original_title', 'size'),
    total_votes=('vote_count', 'sum')
).reset_index()

print("Nhóm dữ liệu theo thuộc tính 'adult', đếm số lượng bộ phim và tổng số lượt vote:")
print(grouped)


```

**B) Nhóm các bộ film theo ngôn ngữ gốc (original_language), tính tổng số bộ film, tổng số lượt vote_count và tính trung của  vote_average .**

```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}

grouped_language = df_movies.groupby('original_language').agg(
    total_movies=('original_title', 'size'),
    total_votes=('vote_count', 'sum'),
    average_vote=('vote_average', 'mean')
).reset_index()

print("Nhóm các bộ phim theo ngôn ngữ gốc (original_language), tính tổng số bộ phim, tổng số lượt vote_count và trung bình vote_average:")
print(grouped_language)


```

Đọc dữ liệu từ file Data_Point.xlsx vào biến kiểu dataframe:

```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}

file_path="/content/drive/MyDrive/for_student/Bai7/ThucHanh/Data_Excercise/Data_Point.xlsx"


df_lop1 = pd.read_excel(file_path, sheet_name=0)

df_lop2 = pd.read_excel(file_path, sheet_name=1, skiprows=9)

df_lop3 = pd.read_excel(file_path, sheet_name=2, header=None)
df_lop3.columns = ["Code", "A", "B1", "B2", "C1", "C2"]

df_lop1.head(), df_lop2.head(), df_lop3.head()
```

### **Yêu cầu 2.2:**
---
Nối 3 DataFrame df_lop1, df_lop2, df_lop3 thành một DataFrame df_full chứa tất cả danh sách bảng điểm của 3 lớp

```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}

df_full = pd.concat([df_lop1, df_lop2, df_lop3], ignore_index=True)
print(df_full.head())

df_full.shape
```

### **Yêu cầu 2.3:**
---
Trong df_full: Tạo một cột Diem_he10 được tính dựa vào các cột tương ứng của từng hàng dữ liệu, theo công thức sau:

**Diem_he10 = 0.6 * A + 0.3 * ((B1+B2)/2) + 0.1 * ((C1+C2)/2)**
    
Làm tròn đến 1 số sau dấu phẩy

```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}

df_full["Diem_he10"] = (
    0.6 * df_full["A"]
    + 0.3 * (df_full["B1"] + df_full["B2"]) / 2
    + 0.1 * (df_full["C1"] + df_full["C2"]) / 2
).round(1)



```

```python
df_full.head()
```

### **Yêu cầu 2.4:**
---
Từ cột Diem_he10 trong df_full tạo một cột Diem_chu, Diem_so theo quy đổi:

    * Điểm hệ 10 Từ 9.0 đến 10.0  ----> Điểm chữ: A+  ----> Điểm số: 4.0
    * Điểm hệ 10 Từ 8.5 đến 8.9  ----> Điểm chữ: A    ----> Điểm số: 3.7
    * Điểm hệ 10 Từ 8.0 đến 8.4  ----> Điểm chữ: B+   ----> Điểm số: 3.5   
    * Điểm hệ 10 Từ 7.0 đến 7.9  ----> Điểm chữ: B    ----> Điểm số: 3.0
    * Điểm hệ 10 Từ 6.5 đến 6.9  ----> Điểm chữ: C+   ----> Điểm số: 2.5  
    * Điểm hệ 10 Từ 5.5 đến 6.4  ----> Điểm chữ: C    ----> Điểm số: 2.0     
    * Điểm hệ 10 Từ 5.0 đến 5.4  ----> Điểm chữ: D+   ----> Điểm số: 1.5
    * Điểm hệ 10 Từ 4.0 đến 4.9  ----> Điểm chữ: D    ----> Điểm số: 1.0  
    * Điểm hệ 10 Từ 0.0 đến 3.9  ----> Điểm chữ: F    ----> Điểm số: 0.0


```python
#{HỌC VIÊN VIẾT CODE VÀO ĐÂY}
def quy_doi_diem(row):
    x = row["Diem_he10"]

    if 9.0 <= x <= 10.0:
        return "A+", 4.0
    elif 8.5 <= x <= 8.9:
        return "A", 3.7
    elif 8.0 <= x <= 8.4:
        return "B+", 3.5
    elif 7.0 <= x <= 7.9:
        return "B", 3.0
    elif 6.5 <= x <= 6.9:
        return "C+", 2.5
    elif 5.5 <= x <= 6.4:
        return "C", 2.0
    elif 5.0 <= x <= 5.4:
        return "D+", 1.5
    elif 4.0 <= x <= 4.9:
        return "D", 1.0
    else:
        return "F", 0.0

df_full["Diem_chu"], df_full["Diem_so"] = zip(*df_full.apply(quy_doi_diem, axis=1))




```

```python
df_full.head()
```

**Mô Tả Bài Toán:**

* Nhằm cải thiện, nâng cao chất lượng phục vụ hành khách. Giúp tìm ra những dịch vụ, những khâu còn hạn chế để đem đến sự hài lòng cho hành khách qua đó thu hút và giữ chân hành khách sử dụng dịch vụ bay của hãng mình.

* Để thực hiện được dự án này, Cần tiến hành lấy khảo sát của tất cả hành khách đã sử dụng dịch vụ bay của hãng thông qua các kênh khác nhau (khảo sát Online, sử dụng các phiếu khảo sát, phỏng vấn ….), thu thập và tổng hợp các dữ liệu này lại để phục vụ phân tích.

**Mô Tả Tập Dữ Liệu:**

Dữ liệu của dự án bao gồm 2 loại:
* File 1: Data_Passenger_Airlines.csv  - Cho biết thông tin về hành khách đi trên các chuyến bay.
* File 2: Data_Feedback_Month.Month.xlsx  - Đánh giá của hành khách về các dịch vụ bay của hãng và cảm nhận của hành khách sau chuyến bay.

*HỌC VIÊN XEM CHI TIẾT MÔ TẢ BÀI TOÁN, TẬP DỮ LIỆU VÀ CÁC YÊU CẦU TRONG SLIDE BÀI GIẢNG*

```python
#{HỌC VIÊN VIẾT CODE XỬ LÝ Ở ĐÂY}

file_1= "/content/drive/MyDrive/for_student/Bai7/ThucHanh/Data_Excercise/Data_Airlines/Data_Passenger_Airlines.csv"
file_2= "/content/drive/MyDrive/for_student/Bai7/ThucHanh/Data_Excercise/Data_Airlines/Data_Feedback_Month.xlsx"

df_passenger = pd.read_csv(file_1)

all_sheets = pd.read_excel(file_2, sheet_name=None)
df_feedback = pd.concat(all_sheets.values(), ignore_index=True)
```

Chuẩn hóa cột Code

```python
df_passenger["Code"] = df_passenger["Code"].astype(str)
df_feedback["Code"] = df_feedback["Code"].astype(str)

```

Loại bỏ trùng lặp và missing values:

```python
df_passenger.drop_duplicates(inplace=True)
df_feedback.drop_duplicates(inplace=True)
df_passenger.dropna(subset=["Code"], inplace=True)
df_feedback.dropna(subset=["Code"], inplace=True)

```

Merge hai DataFrame theo cột Code:

```python
df_full = pd.merge(df_passenger, df_feedback, on="Code", how="inner")
```

Encode cột satisfaction

```python
df_full['satisfaction_num'] = df_full['satisfaction'].apply(lambda x: 1 if x=='satisfied' else 0)
```

Tạo nhóm tuổi (AgeGroup)

```python
bins = [0, 18, 30, 50, 70, 100]
labels = ["<18","18-30","31-50","51-70","71+"]
df_full['AgeGroup'] = pd.cut(df_full['Age'], bins=bins, labels=labels)
```

**Phân tích & trực quan hóa**

1: Dịch vụ được đánh giá tốt / kém nhất

```python
service_cols = ['Inflight wifi service','Departure/Arrival time convenient','Ease of Online booking',
                'Food and drink','Seat comfort','Inflight entertainment','Baggage handling',
                'Checkin service','Inflight service','Cleanliness']

avg_service = df_full[service_cols].mean().sort_values(ascending=False)
print(avg_service)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.barplot(x=avg_service.index, y=avg_service.values)
plt.xticks(rotation=45)
plt.ylabel("Điểm trung bình (1-5)")
plt.title("Đánh giá trung bình từng dịch vụ")
plt.show()
```

2: Nhóm khách hàng hài lòng nhất

```python
print(df_full.groupby('Class')['satisfaction_num'].mean())
print(df_full.groupby('Customer Type')['satisfaction_num'].mean())
print(df_full.groupby('AgeGroup')['satisfaction_num'].mean())
```

3: Tương quan giữa các dịch vụ

```python
plt.figure(figsize=(12,8))
sns.heatmap(df_full[service_cols + ['satisfaction_num']].corr(), annot=True, cmap='coolwarm')
plt.title("Tương quan giữa các dịch vụ và mức độ hài lòng")
plt.show()
```

4: Tỷ lệ tổng quan hài lòng vs không hài lòng

```python
df_full['satisfaction'].value_counts(normalize=True) * 100
sns.countplot(x='satisfaction', data=df_full)
plt.title("Tỷ lệ khách hàng hài lòng / không hài lòng")
plt.show()
```
